```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
print("Đang đọc dữ liệu từ file...")

try:

    df = pd.read_csv('Sample - Superstore.csv', encoding='windows-1252')
    print("Đọc dữ liệu thành công!")
    print(f"Số dòng dữ liệu ban đầu: {df.shape[0]} dòng")

    print("\nĐang tiến hành làm sạch dữ liệu...")

    df = df.dropna()

    df['Order Date'] = pd.to_datetime(df['Order Date'])

    df['Year'] = df['Order Date'].dt.year
    df['Month'] = df['Order Date'].dt.month

    df['Sales'] = df['Sales'].round(2)
    df['Profit'] = df['Profit'].round(2)

    print(" Làm sạch dữ liệu xong!")
    print(f"Số dòng dữ liệu sau khi làm sạch: {df.shape[0]} dòng")

    export_filename = 'Cleaned_Superstore.csv'
    df.to_csv(export_filename, index=False)

    numeric_df = df[['Sales', 'Quantity', 'Discount', 'Profit']]

    plt.figure(figsize=(8, 6))
    sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
    plt.title('Biểu đồ Tương quan (Correlation Heatmap) giữa các chỉ số', fontsize=14, fontweight='bold')
    plt.show()

    print(f"\n THÀNH CÔNG! Đã lưu file dữ liệu sạch với tên: {export_filename}")
    print("Bây giờ bạn hãy tải file này về máy để chuẩn bị cho Power BI nhé!")

    display(df.head(3))

except FileNotFoundError:
    print("LỖI: Không tìm thấy file 'Sample - Superstore.csv'.")
    print("Bạn đã upload file lên Colab chưa? Nhớ kiểm tra lại tên file nhé!")
```
